```python
# Accesses a level 1 flux table for TOI-260 that has both flux and wavelength values

filename = 'dnload_dir_hires_calib1/TOI-260/lev1/tbl/ccd1/flux/HI.20190820.50333_1_01_flux.tbl.gz'

df = pd.read_csv(filename, sep='\s+')

# Isolates the wavelength and flux data from the flux.tbl.gz file
wavelength = df.iloc[:,4]
flux = df.iloc[:,5]
# print(wavelength)
# print(flux)

#print(wavelength[0])

# Empty lists that will be filled by the for loop
wavelength_graph = []
flux_graph = []


# For loop that takes all the values from wavelength and flux and puts them into easy to graph lists.
for i in range(len(wavelength)):
    wavelength_graph.append(wavelength[i])
    flux_graph.append(flux[i])


## Test to see if plotting works. For some reason the astro502 kernel is getting errors when trying to use plt.show()
# x = [1,2,3,4]
# y = [1,2,3,4]

# plt.plot(x,y)
# plt.show()


# To get data plotted, I had to switch to the generic python3 kernel.
plt.plot(wavelength_graph,flux_graph)
plt.xlabel('wavelength (angstroms)')
plt.ylabel('relative flux')
plt.title('TOI-260')
plt.show()





```

```python
# Code that gets the file path for every file in a folder.

directory_path_str = 'dnload_dir_hires_calib1/TOI-260/lev1/tbl/ccd1/flux'

file_paths_os = []
for filename in os.listdir(directory_path_str):
    full_path = os.path.join(directory_path_str, filename)
    if os.path.isfile(full_path):
        file_paths_os.append(full_path)

### Checks to see if the code was successful in getting all of the file paths
#print(file_paths_os)

# Empty lists that will be filled by the for loop
wavelength_full_graph = []
flux_full_graph = []

for i in range(len(file_paths_os)):
    filename = file_paths_os[i]
    
    df = pd.read_csv(filename, sep='\s+')

    wavelength_full = df.iloc[:,4]
    flux_full = df.iloc[:,5]
    # For loop that takes all the values from wavelength and flux and puts them into easy to graph lists.
    for j in range(len(wavelength_full)):
        wavelength_full_graph.append(wavelength_full[j])
        flux_full_graph.append(flux_full[j])


### Testing to make sure the code worked and got all the flux values.
#print(flux_full_graph)

    
plt.plot(wavelength_full_graph,flux_full_graph)
plt.xlabel('wavelength (angstroms)')
plt.ylabel('relative flux')
plt.title('TOI-260')
plt.show()
```
